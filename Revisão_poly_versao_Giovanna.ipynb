### Nascidos Fora

Como os residentes nascidos fora dos Estados Unidos estão distribuídos?

O país, como um todo, tem 13% de residentes nascidos fora, mas nos estados esses percentuais variam de 1.2% na Virgínia Ocidental a 27.2% na Califórnia.

O arquivo NACIDOSFORA.xls apresenta os dados para os 51 estados.

- (a) Faça um histograma e analise brevemente seu resultado. Use classes de largura 5% começando em 0%. Isto é, a primeira barra cobre de 0% a < 5%, a segunda cobre de 5% a < 10%, e assim por diante.
- (b) Dê o resumo dos 5 números da distribuição.
- (c) A Califórnia é um potencial valor atípico ou simplesmente a maior observação da distribuição? O que diz a regra 1.5xAIQ?






```python
# (a)
nf = pd.read_excel('NASCIDOSFORA.xls')

alt.Chart(nf).mark_bar().encode(
    alt.X('PctNascFora:Q', bin=alt.Bin(step=5, extent=[0, 30])),
    y='count()'
)
```

(a) A distribuição é assimétrica à direita (unimodal): a maior parte dos estados concentra-se nas classes baixas (0% a 10%), com uma cauda longa de poucos estados com percentuais bem mais altos, indicando a presença de valores atípicos.

```python
# (b)
nf.PctNascFora.describe()
```

```python
# (c)
q1, q3 = nf.PctNascFora.quantile(0.25), nf.PctNascFora.quantile(0.75)
aiq = q3 - q1                      # AIQ = 12.5 - 3.8 = 8.7
limite_alto = q3 + 1.5 * aiq       # 12.5 + 1.5*8.7 = 25.55

print('Q1:', q1, '| Q3:', q3, '| AIQ:', aiq)
print('Limite superior (Q3 + 1.5*AIQ):', limite_alto)
```

(c) Pela regra do 1,5×AIQ, qualquer valor acima de 25,55% é considerado atípico. Como a Califórnia tem 27,2% — acima desse limite — ela é, sim, classificada como um valor atípico, e não apenas a maior observação 'normal' da distribuição.

O arquivo EMISSAOCO2.xls apresenta as emissões anuais de CO2, por pessoa em 2010 em paises com populações de, no mínimo, 30 milhões naquele ano.

- (a) Por que você acha que escolhemos medir emissões por pessoa em vez do total de emissões de CO2 para cada país?
- (b) Faça um histograma a partir desses dados. Descreva a forma, o centro e a dispersão da distribuição.
- (c) Quais países são valores atípicos?
- (d) Ache a média e a mediana para esses dados. Quais características da distribuição (descrita no item b) explicam a diferença entre a média e a mediana?

(a) Medir o total de CO2 favoreceria automaticamente os países mais populosos (como China ou Índia), mesmo que cada pessoa individualmente polua pouco. Medir per capita permite comparar o impacto médio de cada pessoa, que é uma medida mais justa do padrão de consumo/estilo de vida de cada país, independente do tamanho da população.

```python
# (b)
co2 = pd.read_excel('EMISSAOCO2.xls')

alt.Chart(co2).mark_bar().encode(
    alt.X('CO2:Q', bin=alt.Bin(step=2)),
    y='count()'
)
```

(b)
- Forma: assimétrica à direita (unimodal), com a maioria dos países emitindo pouco e uma cauda de poucos países emitindo muito.

- Centro: mediana ≈ 3,95 toneladas/pessoa (média é maior: 4,61, puxada pelos valores altos).

- Dispersão: vai de 0,03 até 18,9 toneladas — bem espalhado (desvio padrão ≈ 4,73).

```python
# (c)
q1, q3 = co2.CO2.quantile(.25), co2.CO2.quantile(.75)
limite_alto = q3 + 1.5*(q3-q1)

print('Q1:', q1, '| Q3:', q3)
print('Limite superior:', limite_alto)
```

(c) Estados Unidos (18,91) fica acima do limite de 18,73 → é valor atípico. Canadá (16,92) fica logo abaixo do limite, então não é considerado atípico pela regra, mas está bem próximo/alto.

(d) Média = 4,61 | Mediana = 3,95 → a média é maior que a mediana.

Isso acontece porque a distribuição é assimétrica à direita: os poucos países com emissões muito altas (EUA, Canadá) 'puxam' a média para cima, enquanto a mediana — que é resistente a valores extremos — continua representando melhor o país 'típico'.

### Estados Felizes

A felicidade humana, ou bem-estar, pode ser avaliada subjetivamente ou objetivamente. Uma avaliação subjetiva pode ser feita, ouvindo-se o que as pessoas dizem. Uma avaliação objetiva pode ser feita a partir de dados relacionados a bem-estar, como renda, preços das casas, ausência de congestionamentos de tráfego etc. As avaliações subjetivas e objetivas coincidem? Para estudar esse assunto, os investigadores fizeram avaliações, tanto subjetivas como objetivas, para cada um dos 50 estados norte-americanos (que constam no arquivo FELICIDADE.xls).

A medida subjetiva foi o escore médio de questões de satisfação com a vida, encontradas no Behavioral Risk Factor Surveillance System (BRFSS) (Sistema de Avaliação de Risco Comportamental), que é um sistema de pesquisas de saúde com base nos estados. Menores escores indicam um maior grau de felicidade. Para a avaliação objetiva da felicidade, os investigadores calcularam um escore de bem-estar (chamado escore de compensação de diferenciais) para cada estado, com base em medidas objetivas relacionadas à felicidade e ao bem-estar, Os estados foram, então, ordenados de acordo com esse escore (Posto 1 sendo o mais feliz).

- (a) Faça um diagrama de dispersão do BRFSS médio (variável resposta) contra o posto com base nos escores de compensação de diferenciais (variável explicativa).
- (b) Há uma associação geral positiva ou uma associação geral negativa entre BRFSS e o posto com base no método de compensação de diferenciais?
- (c) A associação geral mostra concordância ou discordância entre o escore médio subjetivo BRFSS e o posto com base nos dados objetivos usados no método de compensação de diferenciais?
- (d) Há algum valor atípico? Se for o caso, quais são os escores BRFSS correspondentes a esses valores atípicos?

```python
# (a)
fel = pd.read_excel('FELICIDADE.xls')

alt.Chart(fel).mark_point().encode(
    x=alt.X('CompDif:Q', title='Posto (compensação de diferenciais)'),  # variável explicativa
    y=alt.Y('BRFSS:Q', title='Escore médio BRFSS')                       # variável resposta
)
```

```python
fel.BRFSS.corr(fel.CompDif)
```

(b) correlação = -0,56 → associação negativa (moderada).

(c) Há discordância: estados que se saem melhor na avaliação objetiva (bem-estar material) tendem a reportar menor felicidade subjetiva, e vice-versa. Isso é contraintuitivo — sugere que bem-estar material e felicidade percebida não caminham juntos nesses dados.

```python
# (d)
q1, q3 = fel.BRFSS.quantile(.25), fel.BRFSS.quantile(.75)
aiq = q3 - q1
limite_baixo = q1 - 1.5 * aiq
limite_alto = q3 + 1.5 * aiq

print('Limite inferior:', limite_baixo, '| Limite superior:', limite_alto)

# valores fora desse intervalo seriam os atípicos:
fel[(fel.BRFSS < limite_baixo) | (fel.BRFSS > limite_alto)]
```

(d) nenhum valor de BRFSS ultrapassa os limites (todos entre -0,088 e 0,033) → não há valores atípicos claros nessa variável pela regra do 1,5×AIQ

### Parar de fumar

Um grande experimento aleatorizado foi realizado para avaliar a eficácia de Chantix para parar de fumar, comparado com bupropion (mais comumente conhecido como Wellbutrin ou Zyban) e um placebo. Chantix é diferente da maioria dos produtos para parar de fumar, pois tem como alvo os receptores de nicotina no cérebro, ataca-os e impede que a nicotina os alcance, enquanto o bupropion é um antidepressivo muito usado para ajudar as pessoas a pararem de fumar.

Fumantes com boa saúde geral que fumavam, pelo menos, 10 cigarros por dia, foram associados aleatoriamente aos trata- mentos com Chantix (n = 352), com bupropion (n = 329), ou com o placebo (n = 344). A medida da resposta é a cessação contínua do fumo para as semanas 9 a 12 do estudo.

O arquivo CESSASUMO.xls é uma tabela de dupla entrada dos resultados.

Se um sujeito fumou nas semanas 9 a 12, como isso depende do tratamento recebido?

```python
ces = pd.read_excel('CESSAFUMO.xls')

for tratamento in ['Chantix', 'Bupropion', 'Placebo']:
    grupo = ces[ces.Tratamento == tratamento]           # filtra só esse tratamento
    total = grupo.Contagem.sum()                        # total de pessoas nesse grupo
    continuou_fumando = grupo[grupo.Fumou == 'Sim'].Contagem.sum()  # quantos continuaram fumando

    pct = continuou_fumando / total * 100
    print(tratamento, '-> % que continuou fumando:', pct)
```

obs: a coluna Fumou = Sim significa que a pessoa continuou fumando nas semanas 9-12 (ou seja, o tratamento não funcionou para ela). Fumou = Não significa que ela parou de fumar (tratamento funcionou). Sempre confira os valores da coluna antes de interpretar — é fácil confundir "Sim" com "sucesso".

Há uma associação clara entre tratamento e desfecho: quem tomou Chantix teve a menor taxa de continuar fumando (55,97%), seguido por Bupropion (70,52%), enquanto o grupo Placebo teve a maior taxa de recaída (82,27%). Isso indica que Chantix foi o tratamento mais eficaz para parar de fumar entre os três, seguido por Bupropion, e o Placebo teve o pior desempenho — como era esperado, já que ele não tem princípio ativo algum.

Considere os dados do arquivo InfoFamiliasEntrevistadas.csv, corresponde ao resultado de entrevistas realizadas com uma amostra de 120 famílias de uma certa localidade.

As variáveis que compõem a base são:
- Local - localidade da moradia
- P.a.p. - uso de algum programa de alimentação popular (0 - não; 1 - sim)
- Instr. - último nível de instrução completado pela pessoa com principal renda da casa (1 - nenhum; 2 - fundamental; 3 - médio)
- Tam. - número de pessoas residentes no domicílio
- Renda - renda familiar mensal, em quantidade de salários mínimos



Responda às questões abaixo:

- (a) Classifique as famílias com renda mensal de até 5 salários mínimos, como _renda baixa_ e famílias com rendimentos mensais acima de 5 salários mínimos como _renda alta_.
- (b) A amostra sugere alguma associação entre _renda familiar_ e _uso de programas de alimentação popular_ ? Justifique através da construção e interpretação de uma tabela de dupla entrada.

```python
# (a)
fam = pd.read_csv('InfoFamiliasEntrevistadas.csv')

fam['FaixaRenda'] = 'Renda alta'                          # todo mundo começa como "alta"
fam.loc[fam.Renda <= 5, 'FaixaRenda'] = 'Renda baixa'      # troca pra "baixa" quem tem Renda <= 5

fam.FaixaRenda.value_counts()
```

```python
# (b)
baixa = fam[fam.FaixaRenda == 'Renda baixa']
alta = fam[fam.FaixaRenda == 'Renda alta']

pct_pap_baixa = baixa[baixa['P.a.p.'] == 1].shape[0] / baixa.shape[0] * 100
pct_pap_alta = alta[alta['P.a.p.'] == 1].shape[0] / alta.shape[0] * 100

print('Renda baixa - % que usa PAP:', pct_pap_baixa)
print('Renda alta - % que usa PAP:', pct_pap_alta)
```

(b) Sim, a amostra sugere associação: famílias de renda baixa usam o programa de alimentação popular proporcionalmente mais (72,7%) do que famílias de renda alta (58,5%). Isso é coerente com o esperado, já que programas de alimentação popular são voltados justamente para famílias de menor renda. A diferença de ~14 pontos percentuais indica uma associação moderada entre as duas variáveis — embora ambos os grupos usem o programa em proporção relativamente alta.

```python
# (a)
oecd = pd.read_csv('OECD_IDH_data.csv')

alt.Chart(oecd).mark_bar().encode(
    alt.X('GDP:Q', bin=alt.Bin(step=5000)),
    y='count()'
)
```

(a) A maioria dos países fica entre 19 mil e 33 mil dólares per capita, com um valor bem isolado lá na ponta.

```python
# (b)
q1, q3 = oecd.GDP.quantile(.25), oecd.GDP.quantile(.75)
limite_alto = q3 + 1.5*(q3-q1)   # = 38.479,5

print('Q1:', q1, '| Q3:', q3)
print('Limite superior:', limite_alto)
```

(b) Luxemburgo, com PIB per capita de US$ 69.961, está muito acima do limite (38.479,5) → é claramente um valor atípico.

```python
# (c)
oecd[['Health', 'FemEcon']].describe()
```

(c)
- Health (gasto público em saúde, % do PIB): média 6,7%, varia de 5,1% a 8,8%. Distribuição pouco espalhada.
- FemEcon (atividade econômica feminina, % da masculina): média 76,8%, varia de 61% a 87%.

```python
# (d)
# Célula 1: o número da correlação
print('Correlação:', oecd['Health'].corr(oecd['FemEcon']))
```

```python
# Célula 2: o gráfico de dispersão
alt.Chart(oecd).mark_point().encode(x='Health:Q', y='FemEcon:Q')
```

(d) Países que investem proporcionalmente mais em saúde pública tendem também a ter maior participação econômica feminina relativa. A relação é positiva e moderada — não é uma regra fechada, mas sugere que investimento social mais amplo (saúde) caminha junto com maior equidade de gênero no mercado de trabalho nesses países.

# Questão 14

### Censo Demográfico dos Estados brasileiros em 2000

O arquivo Censo_estados.xlsx possui informações sobre o Cendo Demográfico de 2000, que contém inficadores sociais dos estados brasileiros.

Realize análises bivariadas e interprete-as, para cada par de variáveis.

```python
censo = pd.read_excel('Censo_estados.xlsx')

colunas = ['Esperança de vida ao nascer 2000', 'Mortalidade infantil 2000',
           'Taxa de analfabetismo - 18 anos ou mais de idade 2000', 'Renda per capita 2000']

censo[colunas].corr()   # gera a "matriz" com a correlação de cada par
```

```python
alt.Chart(censo).mark_point().encode(
    x='Renda per capita 2000:Q',
    y='Esperança de vida ao nascer 2000:Q',
    tooltip=['Territorialidades']
)
```

Todos os pares mostram correlações fortes, no sentido esperado: estados com maior renda per capita têm maior esperança de vida, e menor mortalidade infantil e analfabetismo. Isso reflete o padrão clássico de desigualdade regional no Brasil em 2000 — indicadores sociais (saúde, educação) e econômicos (renda) caminham muito juntos: estados mais pobres tendem a apresentar piores indicadores em todas as outras dimensões simultaneamente.
